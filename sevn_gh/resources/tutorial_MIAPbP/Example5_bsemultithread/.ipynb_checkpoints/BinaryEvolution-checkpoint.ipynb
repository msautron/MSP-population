```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("sevn_output_period/output_0.csv")

#Initial parameter
idx=df.BWorldtime==0
print(f"Initial configuration: \n    Mass_primary {df[idx].Mass_0.iloc[0]} Msun Mass_secondary {df[idx].Mass_1.iloc[0]} Msun"
      f"\n    Period: {365.24*df[idx].Period.iloc[0]} days  Eccentricity: {df[idx].Eccentricity.iloc[0]}")

idx=(df.Phase_0==7) & (df.Phase_1==7)
print(f"Final configuration: \n    Mass_primary {df[idx].Mass_0.iloc[0]} Msun Mass_secondary {df[idx].Mass_1.iloc[0]} Msun"
      f"\n    Period: {365.24*df[idx].Period.iloc[0]} days  Eccentricity: {df[idx].Eccentricity.iloc[0]}"
      f"\n    Type primary: {df[idx].RemnantType_0.iloc[0]}   Type secondary: {df[idx].RemnantType_1.iloc[0]}")
```

```python
#Print Period evolution and events

plt.plot(df.BWorldtime, 365.24*df.Period)
plt.xlabel("t [Myr]",fontsize=15)
plt.ylabel("P [days]",fontsize=15)
plt.yscale("log")
plt.show()

plt.plot(df.BWorldtime, df.Mass_0)
plt.plot(df.BWorldtime, df.Mass_1)
plt.xlabel("t [Myr]",fontsize=15)
plt.ylabel("P [days]",fontsize=15)

dfl=df[df.BEvent>0]

i=0
for t,ev in dfl[["BWorldtime","BEvent"]].values:
    
    plt.axvline(t,c="gray",ls="dashed")
    print(ev)
    plt.text(t,3+i,ev)
    i+=1

plt.yscale("log")
plt.show()
```

```python
#HR diagram

plt.scatter(df[df.Phase_0<7].Temperature_0,df[df.Phase_0<7].Luminosity_0)
plt.scatter(df[df.Phase_1<7].Temperature_1,df[df.Phase_1<7].Luminosity_1)

plt.xscale("log")
plt.yscale("log")
plt.xlim(4e5,1e3)
```
