# Example 5

In this example, we evolve a list of  binary systems (in the file [listBin.dat](listBin.dat)) then we run SEVN  in parallel using more than one thread. Finally, we show how to gather the information stored in the outputs of the different threds

- To compile SEVN use the [compile.sh](compile.sh) script
- The run in parappel use [run_multi.sh](run_multi.sh)


Remember to set the proper SEVN path in the compile and run scripts

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Cycle over the output files and concat them 
import glob
outfiles=glob.glob("sevn_output/output_*")

dfs=[]
for outfile in outfiles:
    dfs.append(pd.read_csv(outfile))
    
df=pd.concat(dfs)
display(df)
```

```python
#One can exploit the file splitting to operate in parallel (e.g. filtering)
#Define a function to call in parallel
def filterBCO_df(ifile):
    df=pd.read_csv(ifile)
    df=df[df.Semimajor.notnull() & (df.Phase_0==7) & (df.Phase_1==7)]
    return df

from multiprocess import Pool 

#Use the multiprocess library (pip instal multiprocess) to spawn 3 processes each one working on a separate file
outfiles=glob.glob("sevn_output/output_*")
with Pool(3) as pool:
    dfl=pool.map(filterBCO_df,outfiles)

#Finally generate the final file
df=pd.concat(dfl)
display(df)
```

```python
#Alternative using dask  dataframe (pip install dask, see https://docs.dask.org/en/stable/dataframe.html)
import dask.dataframe as dd

dfd=dd.read_csv("sevn_output/output_*")
display(dfd)
df=dfd.compute()
display(df)
```

```python
#Filter 
dfd=dd.read_csv("sevn_output/output_*")
dfd=dfd[dfd.Semimajor.notnull() & (dfd.Phase_0==7) & (dfd.Phase_1==7)]
display(dfd)
df=dfd.compute()
display(df)
```
